```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
from cyvlfeat import sift
from sift import *
from harris import *
```

```python
imname = '../data/empire.jpg'
im1 = np.array(Image.open(imname).convert('L'))

plt.figure()
plt.gray()
plt.imshow(im1)
```

```python
l1 = sift.sift(im1, edge_thresh=10, peak_thresh=5)

harrisim = compute_harris_response(im1)
filtered_coords = get_harris_points(harrisim, 10)
```

```python
plot_features(im1, l1, circle=False)
```

```python
plot_features(im1, l1, circle=True)
```

```python
plot_harris_points(im1, filtered_coords)
```
