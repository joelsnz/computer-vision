```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.ndimage import gaussian_filter
from scripts.rof import denoise
from PIL import Image
```

```python
# create synthetic image with noise

im = np.zeros ((500, 500))
im[100:400, 100:400] = 128
im[200:300, 200:300] = 255
im = im + 30 * np.random.standard_normal((500, 500))

Image.fromarray(np.uint8(im))
```

```python
U, T = denoise(im, im)
G = gaussian_filter(im, 10)
```

```python
Image.fromarray(np.uint8(U))
```

```python
Image.fromarray(np.uint8(G))
```

```python
im = np.array(Image.open('Resources/bird.jpg').convert('L'))
U, T = denoise(im, im)

plt.figure()
plt.gray()
plt.imshow(U)
plt.axis('equal')
plt.axis('off')
plt.show()
```
