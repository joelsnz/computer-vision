# Graylevel transforms

```python
im = np.array(Image.open('Resources/bird.jpg').convert('L'))

im2 = 255 - im # invert image

im3 = (100.0/255) * im + 100 # clamp to interval 100...200

im4 = 255.0 * (im/255.0)**2 # squared (lowers the values of darker pixels)
```

```python
pil_im = Image.fromarray(np.uint8(im4))
pil_im
```

```python
im = np.array(Image.open('Resources/bird.jpg').convert('L'))
im2, cdf = histeq(im)
```

```python
plt.figure()
plt.hist(im2.flatten(), 256)
plt.show()
```
