```python
from PIL import Image
```

```python
# create a PIL image object
pil_im = Image.open('Resources/bird.jpg')
```

```python
# convert the image to grayscale and save it
pil_im.convert('L').save('Resources/bird_grayscale.jpg')
```

```python
# cropping a region of the image
box = (100, 100, 800, 800)
region = pil_im.crop(box)
```

```python
# rotate the region and paste it to the original image
region = region.transpose(Image.ROTATE_180)
pil_im.paste(region, box)
```

```python
# resize the image
out = pil_im.resize((720, 480))
```

```python
# rotate the image counter clockwise
out = out.rotate(45)
```
